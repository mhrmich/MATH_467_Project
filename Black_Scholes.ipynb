# Black-Scholes Pricing

This notebook demonstrates the Black-Scholes closed-form formulas for pricing European call
and put options, alongside their key Greeks. The implementations rely on the
`black_scholes` helper function defined in the accompanying `black_scholes.py` module, which
uses the standard normal cumulative distribution function computed via `math.erf`.


```python
from pprint import pprint

from black_scholes import black_scholes
```

## Example parameters

The parameters below mirror a classic textbook example (at-the-money option with one year to
expiry, 5% risk-free rate, and 20% volatility). Feel free to modify these values and rerun
the cell to explore how prices and sensitivities respond.


```python
parameters = {
    "spot": 100.0,
    "strike": 100.0,
    "time_to_maturity": 1.0,
    "risk_free_rate": 0.05,
    "volatility": 0.2,
}

call_result = black_scholes("call", **parameters)
put_result = black_scholes("put", **parameters)

print("Inputs:")
pprint(parameters)

print("\nCall result:")
pprint(call_result)

print("\nPut result:")
pprint(put_result)
```
